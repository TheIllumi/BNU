```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
        
df = pd.read_csv('/kaggle/input/houseing/Housing.csv')
```

```python
df.head(), df.info()
```

```python
# Define features and target variable
x = df[['area']]
y = df['price']

# Split the data into training and testing sets
x_train = x[:500]
x_test = x[500:]
y_train = y[:500]
y_test = y[500:]

x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.linear_model import LinearRegression

# Initialize the Linear Regression model
model = LinearRegression()

# Train the model
model.fit(x_train, y_train)

# Coefficients and intercept of the trained model
model.coef_, model.intercept_
```

```python
from sklearn.metrics import mean_squared_error

# Predict house prices for the test set
yhat = model.predict(x_test)

# Calculate the Mean Squared Error (MSE)
mse = mean_squared_error(y_test, yhat)
mse
```

```python
from sklearn.metrics import mean_squared_error

# Predict house prices for the test set
yhat = model.predict(x_test)

# Calculate the Mean Squared Error (MSE)
mse = mean_squared_error(y_test, yhat)
print("Mean Squared Error for model with 'area' as feature:", mse)
```

```python
# Using 'area' and 'bedrooms' as features
x = df[['area', 'bedrooms']]

# Split into training and testing sets
x_train = x[:500]
x_test = x[500:]

# Train the model and calculate MSE
model.fit(x_train, y_train)
yhat = model.predict(x_test)
mse = mean_squared_error(y_test, yhat)
print("Mean Squared Error for model with 'area' and 'bedrooms':", mse)
```

```python
# Define different feature sets
features_list = [
    ['area'],  # One feature
    ['area', 'bedrooms'],  # Two features
    ['area', 'bedrooms', 'bathrooms'],  # Three features
    ['area', 'bedrooms', 'bathrooms', 'stories'],  # Four features
    ['area', 'bedrooms', 'bathrooms', 'stories', 'parking'],  # Five features
]

# Loop through each feature set and calculate the MSE
results = []

for features in features_list:
    # Select the current set of features
    x = df[features]
    
    # Split the data into training and testing sets
    x_train = x[:500]
    x_test = x[500:]
    y_train = y[:500]
    y_test = y[500:]
    
    # Initialize and train the model
    model = LinearRegression()
    model.fit(x_train, y_train)
    
    # Predict house prices and calculate the MSE
    yhat = model.predict(x_test)
    mse = mean_squared_error(y_test, yhat)
    
    # Store the results
    results.append({'Features': features, 'MSE': mse})

# Display the results
for result in results:
    print(f"Features: {result['Features']} -> MSE: {result['MSE']}")
```
